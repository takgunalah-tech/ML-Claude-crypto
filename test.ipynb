```python
import yfinance as yf
import pandas as pd

def test_ticker(symbol):
    print(f"--- Testing Ticker: {symbol} ---")
    try:
        # Download data
        data = yf.download(symbol, period="3d", interval="1h", progress=False)

        if data.empty:
            print(f"Result: No data found for {symbol}.\n")
            return

        # Check for Timezone
        is_tz_aware = data.index.tz is not None
        tz_name = data.index.tz if is_tz_aware else "None (Naive)"
        print(f"Data found! Rows: {len(data)}")
        print(f"Timezone: {tz_name}")

        # FIX: Access the 'Close' column safely even if it's a MultiIndex
        # We use .iloc[-1] on the specific column to get a single float value
        if isinstance(data['Close'], pd.DataFrame):
            last_price = data['Close'][symbol].iloc[-1]
        else:
            last_price = data['Close'].iloc[-1]

        print(f"Latest Price: {last_price:.4f}")
        print(f"Last timestamp: {data.index[-1]}")
        
    except Exception as e:
        print(f"An error occurred: {e}")
    print("-" * 30 + "\n")


```

```python

# Re-run your tests 

tickers = ["^GSPC", "QQQ", "GC=F", "TLT", "DX-Y.NYB"]
for t in tickers:
    test_ticker(t)
```
